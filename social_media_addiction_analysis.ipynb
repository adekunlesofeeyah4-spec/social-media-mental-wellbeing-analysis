## 2. Question 1A - What age group has the highest and lowest average mental_wellbeing_score?
##    Question 1B - What are the most and least common occupation-relationship status combination among the age groups with the highest and lowest average mental_wellbeing_score?

### 2.1 Creating Age Groups

```python
df['Age_Group'] = pd.cut(df['Age'], bins = [15, 20, 25, 30, 35, 40, 45],
labels = ['Adolescence/Late Teens', 'Early Adulthood', 'Later 20s', 'Early 30s', 'Later 30s', 'Early 40s'], right=False)
```

### 2.2 Question 1A - Analysis

```python
df.groupby('Age_Group', observed=True)['Mental_Wellbeing_Score'].agg(['count', 'mean', 'median', 'min', 'max'])
```

### Question 1A - Visualization


```python
df.boxplot(column='Mental_Wellbeing_Score', by='Age_Group', figsize = (8,4))
```

### 2.4 Question 1B - Analysis

```python
filtered = df[df['Age_Group'].isin(['Early 30s', 'Early 40s'])]
filtered
```

```python
counts = filtered.groupby(['Age_Group', 'Occupation', 'Relationship_Status'], observed=True).size()

counts / counts.groupby(level=['Age_Group'], observed=True).sum() * 100
```

### 5.1 Question 4A - Analysis


```python
df.groupby('Gender')['Mental_Wellbeing_Score'].agg(['count', 'mean', 'median', 'min', 'max'])
```

### 5.2 Question 4B - Analysis


```python
df['Mental_Wellbeing_Score'].median()
```

```python
df['Wellbeing_Group'] = np.where(df['Mental_Wellbeing_Score'] < 6.40, 'Lower_Wellbeing', 'Not_Lower_Wellbeing')

df['Wellbeing_Group'].value_counts()
```

### 5.3 Categorical Characteristics

#### Addiction Level


```python
add_counts = df.groupby(['Gender', 'Addiction_Level', 'Wellbeing_Group']).size()
lower_add_counts = add_counts[add_counts.index.get_level_values('Wellbeing_Group') == 'Lower_Wellbeing']
lower_add_counts / lower_add_counts.groupby(level=['Gender'], observed=True).sum() * 100
```

#### Occupation


```python
occ_counts = df.groupby(['Gender', 'Occupation', 'Wellbeing_Group']).size()
lower_occ_counts = occ_counts[occ_counts.index.get_level_values('Wellbeing_Group') == 'Lower_Wellbeing']
lower_occ_counts / lower_occ_counts.groupby(level=['Gender'], observed=True).sum() * 100
```

#### Relationship Status


```python
rel_counts = df.groupby(['Gender', 'Relationship_Status', 'Wellbeing_Group']).size()
lower_rel_counts = rel_counts[rel_counts.index.get_level_values('Wellbeing_Group') == 'Lower_Wellbeing']
lower_rel_counts / lower_rel_counts.groupby(level=['Gender'], observed=True).sum() * 100
```
